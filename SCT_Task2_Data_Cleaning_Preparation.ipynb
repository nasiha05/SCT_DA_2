# Task 2 – Data Cleaning and Preparation

This notebook focuses on cleaning and preparing the Global Superstore dataset for further analysis.

The main data preparation steps include:

- Loading and inspecting the dataset
- Checking for missing values
- Checking and removing duplicate records
- Removing non-informative columns
- Standardizing column names and text fields
- Converting date columns to the correct data type
- Validating date and numerical fields
- Performing final data-quality checks
- Exporting the cleaned dataset

## 1. Import Required Libraries

The required Python libraries are imported for data loading, cleaning, and analysis.

```python
import pandas as pd
import numpy as np

try:
    from google.colab import files
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# Display settings
pd.set_option('display.max_columns', None)

print("Libraries imported successfully.")
```

## 2. Load the Dataset

The Global Superstore CSV file is uploaded and loaded into a Pandas DataFrame.

```python
if IN_COLAB:
    uploaded = files.upload()
    file_name = next(iter(uploaded))
else:
    file_name = "superstore.csv"

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File name:", file_name)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 3. Initial Data Inspection

The first few records, dataset dimensions, column names, and data types are checked to understand the structure of the raw dataset.

```python
print("First 5 rows:")
display(df.head())

print("\nDataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Data types:")
print(df.dtypes)
```

## 4. Missing Value Check

Missing values are checked for every column. The number and percentage of missing values are displayed only for columns that contain missing data.

```python
missing_count = df.isnull().sum()

missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': (missing_count / len(df) * 100).round(2)
})

missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

if missing_summary.empty:
    print("No missing values were found.")
else:
    display(missing_summary)
```

### Handling Missing Values

If missing values are present, numerical columns are filled using their median values and categorical columns are filled using their most frequent value.

If no missing values are found, no changes are made.

```python
numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include='object').columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        mode_value = df[column].mode()
        if not mode_value.empty:
            df[column] = df[column].fillna(mode_value[0])

print("Missing-value handling completed.")
print("Remaining missing values:", df.isnull().sum().sum())
```

## 5. Duplicate Row Check

Duplicate records are checked to identify repeated rows in the dataset.

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows found:", duplicate_count)
```

### Removing Duplicate Rows

Any completely duplicated records are removed to avoid repeated observations in the cleaned dataset.

```python
rows_before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

rows_after = len(df)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
```

## 6. Removing Non-Informative Columns

Columns that contain the same single value in every row (or are always empty) do not provide any useful information for analysis and are removed.

```python
constant_columns = [col for col in df.columns if df[col].nunique(dropna=False) <= 1]

if constant_columns:
    print("Constant / non-informative columns found:", constant_columns)
    df = df.drop(columns=constant_columns)
    print("Columns removed. New shape:", df.shape)
else:
    print("No constant columns found.")
```

## 7. Standardize Column Names

Column names are cleaned by removing unnecessary spaces from the beginning and end of each name.

```python
df.columns = df.columns.str.strip()

print("Column names standardized successfully.")
```

## 8. Standardize Text Fields

Text fields are cleaned by removing unnecessary leading and trailing spaces.

This helps maintain consistent categorical values during later analysis.

```python
for column in df.columns:
    if df[column].dtype == 'object':
        df[column] = df[column].str.strip()

print("Text fields standardized successfully.")
```

## 9. Convert Date Columns

The order date and ship date columns are converted from text to the datetime data type. This allows the dates to be used correctly for time-based analysis.

```python
# Identify the date columns regardless of whether they use spaces or dots (e.g. "Order.Date")
date_columns = [column for column in df.columns
                if column.lower().replace(" ", "_").replace(".", "_") in ["order_date", "ship_date"]]

print("Date columns found:", date_columns)
```

```python
for column in date_columns:
    df[column] = pd.to_datetime(df[column], errors='coerce')

print("Date conversion completed.")
for column in date_columns:
    print(f"{column}: {df[column].dtype}")
```

## 10. Date Validation

A basic validation check is performed to ensure that the shipping date is not earlier than the order date.

```python
invalid_dates = df[df['Ship.Date'] < df['Order.Date']]

print("Rows where Ship Date is before Order Date:", len(invalid_dates))
```

## 11. Numeric Data Validation

The main numerical fields are checked for invalid or unexpected values.

The checks include:

- Sales less than or equal to zero
- Quantity less than or equal to zero
- Discount below 0
- Discount above 1

```python
checks = {
    "Sales <= 0": (df['Sales'] <= 0).sum(),
    "Quantity <= 0": (df['Quantity'] <= 0).sum(),
    "Discount < 0": (df['Discount'] < 0).sum(),
    "Discount > 1": (df['Discount'] > 1).sum()
}

for check, result in checks.items():
    print(f"{check:18}: {result}")
```

### Handling Invalid Numeric Records

Rows with non-positive Sales or Quantity, or a Discount value outside the valid 0–1 range, do not represent valid transactions and are removed from the dataset.

```python
rows_before_numeric = len(df)

df = df[
    (df['Sales'] > 0) &
    (df['Quantity'] > 0) &
    (df['Discount'] >= 0) &
    (df['Discount'] <= 1)
].reset_index(drop=True)

rows_after_numeric = len(df)

print("Rows before:", rows_before_numeric)
print("Rows after:", rows_after_numeric)
print("Invalid rows removed:", rows_before_numeric - rows_after_numeric)
```

## 12. Additional Useful Columns

A few simple derived columns are added to make the cleaned dataset more useful for further analysis.

- Order Year
- Order Month
- Shipping Days
- Profit Margin

```python
df['Order Year'] = df['Order.Date'].dt.year
df['Order Month'] = df['Order.Date'].dt.month
df['Shipping Days'] = (df['Ship.Date'] - df['Order.Date']).dt.days
df['Profit Margin'] = np.where(
    df['Sales'] != 0,
    df['Profit'] / df['Sales'],
    0
)

print("Additional columns created successfully.")
print("New columns: Order Year, Order Month, Shipping Days, Profit Margin")
```

## 13. Final Validation

The cleaned dataset is checked again to confirm that missing values and duplicate rows have been handled and that the expected data types are present.

```python
remaining_missing = df.isnull().sum().sum()
remaining_duplicates = df.duplicated().sum()

print("Final dataset shape:", df.shape)
print("Remaining missing values:", remaining_missing)
print("Remaining duplicate rows:", remaining_duplicates)

print("\nDate columns:")
for column in date_columns:
    print(f"{column}: {df[column].dtype}")
```

## 14. Export Cleaned Dataset

The cleaned dataset is exported as a CSV file so that it can be used for further analysis and reporting.

```python
output_file = "Global_Superstore_Cleaned.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset exported successfully!")
print("File name:", output_file)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```
